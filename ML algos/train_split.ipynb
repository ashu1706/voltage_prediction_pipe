```python
df=pd.read_excel('Data_file.ods', engine='odf')
```

```python
train_df = pd.DataFrame()
test_df = pd.DataFrame()
```

```python
for i in range(0, 6):
    interval_data = df[(df['T'] >= i) & (df['T'] < i+1)]
    train_split, test_split = train_test_split(interval_data, test_size=0.2, random_state=42)

    train_filename = f'train_interval_{i}_to_{i+1}_seconds.csv'
    test_filename = f'test_interval_{i}_to_{i+1}_seconds.csv'

    train_split.to_csv(train_filename, index=False)
    test_split.to_csv(test_filename, index=False)

    print(f'Training and testing data for {i}-{i+1} seconds interval saved.')
     
     
```
